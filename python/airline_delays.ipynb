# Correct and verify data format transformed

```python
raw_folder = Path(r"D:\Data_Projects\airline_delays_project\data\raw")

for month in (1, 2, 3):
    file = raw_folder / f"flights_2024_{month:02d}.csv"

    row_count = 0
    wrong_month = 0
    invalid_dates = 0
    earliest_date = None
    latest_date = None

    for chunk in pd.read_csv(
        file,
        usecols=["YEAR", "MONTH", "FL_DATE"],
        chunksize=100_000
    ):
        
        dates = pd.to_datetime(
            chunk["FL_DATE"],
            format="%m/%d/%Y %I:%M:%S %p",
            errors="coerce"
        )
        row_count += len(chunk)
        invalid_dates += int(dates.isna().sum())

        correct_period = (
            chunk["YEAR"].eq(2024)
            & chunk["MONTH"].eq(month)
            & dates.dt.year.eq(2024)
            & dates.dt.month.eq(month)
        )
        wrong_month += int((~correct_period).sum())

        valid_dates = dates.dropna()

        if not valid_dates.empty:
            chunk_min = valid_dates.min()
            chunk_max = valid_dates.max()

            earliest_date = (
                chunk_min if earliest_date is None
                else min(earliest_date, chunk_min)
            )
            latest_date = (
                chunk_max if latest_date is None
                else max(latest_date, chunk_max)
            )

    print(f"\nFile: {file.name}")
    print(f"Total rows: {row_count:,}")
    print(f"Date range: {earliest_date} → {latest_date}")
    print(f"Invalid dates: {invalid_dates:,}")
    print(f"Rows outside expected period: {wrong_month:,}")
```

# Create the connection

```python
from pathlib import Path
from getpass import getpass

import pandas as pd
from sqlalchemy import create_engine, text
from sqlalchemy.engine import URL

raw_folder = Path(r"D:\Data_Projects\airline_delays_project\data\raw")

username = input("PostgreSQL username: ").strip()
password = getpass("PostgreSQL password: ")

connection_url = URL.create(
    "postgresql+psycopg2",
    username=username,
    password=password,
    host="localhost",
    port=5432,
    database="airline_delays"
)

engine = create_engine(connection_url)
del password

with engine.connect() as connection:
    print("Connected to:", connection.execute(
        text("SELECT current_database()")
    ).scalar_one())
```

# Load the three CSVs into airline.raw_flights

```python
text_columns = [
    "OP_UNIQUE_CARRIER", "TAIL_NUM", "OP_CARRIER_FL_NUM",
    "ORIGIN", "ORIGIN_CITY_NAME", "ORIGIN_STATE_ABR",
    "DEST", "DEST_CITY_NAME", "DEST_STATE_ABR",
    "CRS_DEP_TIME", "DEP_TIME", "CRS_ARR_TIME", "ARR_TIME",
    "CANCELLATION_CODE"
]

expected_counts = {
    1: 547_271,
    2: 519_221,
    3: 591_767
}

with engine.begin() as connection:
    existing_rows = connection.execute(
        text("SELECT COUNT(*) FROM airline.raw_flights")
    ).scalar_one()

    if existing_rows != 0:
        raise ValueError(
            f"Import stopped: raw_flights already contains "
            f"{existing_rows:,} rows."
        )

    expected_headers = None

    for month, expected_count in expected_counts.items():
        file = raw_folder / f"flights_2024_{month:02d}.csv"
        loaded = 0

        for chunk in pd.read_csv(
            file,
            dtype={column: "string" for column in text_columns},
            chunksize=50_000,
            low_memory=False
        ):
            headers = chunk.columns.tolist()

            if expected_headers is None:
                expected_headers = headers

            if headers != expected_headers or len(headers) != 41:
                raise ValueError(f"Unexpected columns in {file.name}")

            dates = pd.to_datetime(
                chunk["FL_DATE"],
                format="%m/%d/%Y %I:%M:%S %p",
                errors="raise"
            )

            correct_period = (
                chunk["YEAR"].eq(2024)
                & chunk["MONTH"].eq(month)
                & dates.dt.year.eq(2024)
                & dates.dt.month.eq(month)
            )

            if not correct_period.all():
                raise ValueError(f"Unexpected period in {file.name}")

            chunk["FL_DATE"] = dates.dt.date
            chunk.columns = chunk.columns.str.lower()

            chunk.to_sql(
                name="raw_flights",
                con=connection,
                schema="airline",
                if_exists="append",
                index=False,
                chunksize=1_000,
                method="multi"
            )

            loaded += len(chunk)

        if loaded != expected_count:
            raise ValueError(
                f"{file.name}: expected {expected_count:,} rows, "
                f"found {loaded:,}."
            )

        print(f"{file.name}: {loaded:,} rows staged")

print("Import committed successfully.")
```
